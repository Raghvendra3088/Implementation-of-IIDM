```python
import os

print(" Verifying Raw Data Paths on IITP Server...")
base_path = "data/raw"
required_folders = ['sentinel2', 'alos_dem', 'eth_canopy', 'gf1']

all_good = True
for folder in required_folders:
    path = os.path.join(base_path, folder)
    if os.path.exists(path) and len(os.listdir(path)) > 0:
        print(f" Found data in: {path}")
    else:
        print(f" MISSING DATA in: {path}")
        all_good = False

if all_good:
    print("\n All raw data paths are perfectly connected!")
else:
    raise Exception("Stop! Data is missing in the folders above.")

#  tools just in case ml.env missed them
!pip install wandb rasterio scikit-image tqdm matplotlib pandas
```
